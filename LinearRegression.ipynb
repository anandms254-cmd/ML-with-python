```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


data = {
    "Load (N)": [5, 10, 15, 20, 25],
    "Extension (mm)": [1.2, 2.5, 3.7, 4.9, 6.1]}


df = pd.DataFrame(data)


print(df)


x = df[["Load (N)"]]
y = df["Extension (mm)"]


model = LinearRegression()


model.fit(x, y)


prediction = model.predict([[55]])


print("Slope (m):", model.coef_[0])
print("Intercept (c):", model.intercept_)

print(f"Regression Equation: y = {model.coef_[0]:.3f}x + {model.intercept_:.2f}")

print("Predicted extension for 55 N:", prediction[0])


plt.scatter(x, y)


plt.plot(x, model.predict(x))


plt.xlabel("Load (N)")
plt.ylabel("Extension (mm)")
plt.title("Linear Regression")


plt.show()
```
